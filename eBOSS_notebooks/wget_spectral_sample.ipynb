```python
wget_strings =[]
df = pd.DataFrame()

for row in range(len(high_ion_eBOSS)):
    plate = str(high_ion_eBOSS['PLATE'][row])
    pmf = str(high_ion_eBOSS['PMF_STRING'][row])[5:]

    string = plate + '/spec-' + plate + pmf + '.fits'
    wget_strings.append(string)

df['STRING'] = wget_strings
df.to_csv('fits_files/wget_strings.txt', index=False, header=False)
```
