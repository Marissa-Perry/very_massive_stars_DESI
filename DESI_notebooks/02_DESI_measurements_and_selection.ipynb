## Load in DESI parent sample

wavlengths are in units of Angstroms and fluxes are in units of ergs/cm^2/s/Angstromn

```python
filename = 'DESI_parent_catalog.pkl'
path = os.path.join('..', 'subsamples', filename)  # path to dataframe

desi_sample = pd.read_pickle(path)

# ----- adjust units of data -----
scale = 1e-17  # DESI flux unit: 1e-17 [erg/s/cm^2/A]
desi_sample['flux'] = desi_sample['flux'] * scale
desi_sample['flux_ivar'] = desi_sample['flux_ivar'] / scale**2

def ivar_to_err(ivar):
    ivar = np.array(ivar)
    err = np.full_like(ivar, np.nan)     
    good = ivar > 0   # avoid dividing by zero
    err[good] = 1.0 / np.sqrt(ivar[good])
    return err

desi_sample['flux_err'] = desi_sample['flux_ivar'].apply(ivar_to_err)
# ---------------------------------

desi_sample.head()
```
